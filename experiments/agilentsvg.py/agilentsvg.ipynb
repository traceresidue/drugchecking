# Agilent chromatogram → SVG

Export a ChemStation or MassHunter acquisition chromatogram to an annotated SVG.

**Valid input:** a `*.D` acquisition folder (with `.ch` or `AcqData/`) or a single `.ch` file.

**Not valid:** NIST/MassHunter spectral library `.L` folders or their `FULL.D` files — those store reference spectra, not run chromatograms.

## 2. Set paths

Edit the two paths below, then run the rest of the notebook.

```python
from pathlib import Path

# ChemStation *.D folder, MassHunter *.D folder, or a single *.ch file
INPUT_PATH = Path(r"B:\path\to\your\SAMPLE.D")

# Where to write the SVG
OUTPUT_SVG = Path(r"B:\path\to\output\chromatogram.svg")

# Leave None to label peaks above 5% of max intensity
BASELINE_THRESHOLD = None
```

## 3. Load chromatogram

```python
import sys

NOTEBOOK_DIR = Path.cwd()
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

from agilentsvg import load_agilent_chromatogram, render_chromatogram_figure

time_minutes, abundance, channel = load_agilent_chromatogram(INPUT_PATH)
title_name = INPUT_PATH.name if INPUT_PATH.is_dir() else INPUT_PATH.parent.name

print(f"Loaded {len(time_minutes):,} points from channel {channel!r}")
print(f"Retention time range: {time_minutes.min():.3f} – {time_minutes.max():.3f} min")
print(f"Max intensity: {abundance.max():,.0f}")
```

## 4. Preview

```python
%matplotlib inline

import matplotlib.pyplot as plt

fig, ax, peak_labels = render_chromatogram_figure(
    time_minutes,
    abundance,
    title_name,
    baseline_threshold=BASELINE_THRESHOLD,
)
print(f"Marked {len(peak_labels)} peaks")
plt.show()
```

## 5. Export SVG

```python
OUTPUT_SVG.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(OUTPUT_SVG, format="svg", bbox_inches="tight", transparent=True)
print(f"Saved: {OUTPUT_SVG.resolve()}")
```
